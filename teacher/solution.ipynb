## 我的關卡

`course` 是一個清單，遊戲會照順序把每一項放到跑道上，全部跑完就會看到球門。

| 名稱 | 是什麼 | 怎麼過 |
| --- | --- | --- |
| `'cone'` | 三角錐 | 跳過去 |
| `'defender'` | 鏟球的防守球員 | 跳過去 |
| `'shoe'` | 從頭頂飛過來的球鞋 | 站著不要跳 |
| `'ball'` | 地上的足球 | 撿起來可以加分 |
| `'gap'` | 空白 | 什麼都沒有，可以喘口氣 |

關卡用迴圈來排：

- `course.append('cone')` 會在清單最後面加上一個 `'cone'`。
- `for i in range(5):` 會把底下縮排的那幾行重複執行 5 次。
- 所以迴圈裡寫幾行 `append`，每跑一圈就加幾個東西。

```python
# ===== 我的關卡 =====
# course 一開始是空的清單，下面的迴圈每跑一圈，就往清單最後面加幾個東西。
course = []

# range(5) 會讓迴圈跑 5 次，每一次加一組「三角錐、空白、足球」。
# 跑完之後 course 總共有 5 × 3 = 15 個東西。想讓關卡更長，就把 5 改大；
# 想換花樣，就改迴圈裡 append 的名稱，或多寫幾行 append。
for i in range(5):
	course.append('cone')
	course.append('gap')
	course.append('ball')
```

## 遊戲規則

下面的函式不是由你呼叫，而是遊戲在特定時機**自動呼叫**的。每個函式都會收到 `game`，代表整個遊戲，你可以讀取或修改它的這些變數：

| 變數 | 意思 |
| --- | --- |
| `game.on_ground` | 球員是不是站在地上（`True` 或 `False`） |
| `game.vy` | 上下的速度，負數往上、正數往下 |
| `game.lives` | 剩下幾條命 |
| `game.speed` | 往前跑的速度 |
| `game.over()` | 呼叫它，遊戲就結束 |

```python
# ===== 遊戲規則 =====
jump_power = 12

# 按下跳躍鍵時會執行這裡
def on_jump_key(game):
	# 問題二的答案是 if game.on_ground:，只能在地上跳。
	# 這裡寫的是挑戰一（二段跳）的答案：跳了不到 2 次都可以再跳，落地時 jump_count 會自動變回 0
	if game.jump_count < 2:
		game.vy = -jump_power
		game.jump_count = game.jump_count + 1
```

```python
# 撞到障礙時會執行這裡
def on_hit(game):
	game.lives = game.lives - 1
	if game.lives == 0:
		game.over()
```

```python
# 每過一秒會執行這裡
def on_tick(game):
	game.speed = game.speed + 0.2
```

### 挑戰二：射門

撿到地上的足球會增加一發彈藥，按 F 可以射門，把防守球員踢飛！

- 在下面那一格寫一個新的函式 `on_shoot_key(game)`，按 F 或 X 時遊戲會執行它。
- `game.ammo` 是彈藥數，有彈藥才能射門，射完要減一。
- 呼叫 `game.shoot()` 就會射出一顆球。

```python
# ===== 挑戰二：射門 =====
# 在這裡寫 on_shoot_key(game)，寫好之後重新執行這一格
def on_shoot_key(game):
	if game.ammo > 0:
		game.shoot()
		game.ammo = game.ammo - 1
```
