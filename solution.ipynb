## 遊戲規則

下面的函式不是由你呼叫，而是遊戲在特定時機**自動呼叫**的。每個函式都會收到 `game`，代表整個遊戲，你可以讀取或修改它的這些變數：

| 變數 | 意思 |
| --- | --- |
| `game.on_ground` | 球員是不是站在地上（`True` 或 `False`） |
| `game.vy` | 上下的速度，負數往上、正數往下 |
| `game.lives` | 剩下幾條命 |
| `game.speed` | 往前跑的速度 |
| `game.over()` | 呼叫它，遊戲就結束 |

```python
# ===== 遊戲規則 =====
jump_power = 12

# 按下跳躍鍵時會執行這裡
def on_jump_key(game):
	if game.on_ground:
		game.vy = -jump_power
```

```python
# 撞到障礙時會執行這裡
def on_hit(game):
	game.lives = game.lives - 1
	if game.lives == 0:
		game.over()
```

```python
# 每過一秒會執行這裡
def on_tick(game):
	game.speed = game.speed + 0.2
```
